## Градиентный спуск: функции одной переменой

```
def func(x):
    return x**3 - 6*x**2 - 15*x + 10
```

```
x = np.linspace(-5, 10, 150)
```

```
fig = plt.figure()
fig.set_size_inches(10, 8)
plt.plot(x, func(x))
```

```
def func_derivative(x):
    return 3*x**2 - 12*x - 15
```

```
def descent_and_plot(X_start=10, learning_rate=0.01, threshold=0.001, ax=None):
    # создаём figure/axes только на первом вызове
    if ax is None:
        fig, ax = plt.subplots(figsize=(10, 8))
        ax.plot(x, func(x), alpha=0.5)
        ax.set_xlabel("x")
        ax.set_ylabel("f(x)")

    X_next = X_start - learning_rate * func_derivative(X_start)

    # рисуем текущий шаг
    ax.scatter([X_start, X_next], [func(X_start), func(X_next)], color='green', s=80)
    ax.plot([X_start, X_next], [func(X_start), func(X_next)], color='red')

    # условие выхода: если шаг маленький — стоп
    if abs(X_start - X_next) <= threshold:
        return X_next
    else:
        return descent_and_plot(X_next, learning_rate, threshold, ax=ax)

```

```
minimum  = descent_and_plot(10, 0.01, 0.0001)
print(f'Найденная точка в конце градиентного спуска это X={minimum}')
```
