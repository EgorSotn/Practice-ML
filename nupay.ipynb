```
X = np.array([[23, 0.5, 1],
              [35, 1, 1],
              [18, 0, 1]])
Y = np.array([55, 100, 45])
```

```
xxt = np.dot(X.T, X)
xxt
```

```
xxt_inv = np.linalg.inv(xxt)
xxt_inv
```

```
xxt_inv_xxt =  np.dot(xxt_inv, X.T)
xxt_inv_xxt
```

```
final_beta = np.dot(xxt_inv_xxt, Y)
final_beta
```
